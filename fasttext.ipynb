```python
import pandas as pd

from src.utils import PROCESSED_DATA_DIR

from src.features import (
    split_data,
    train_fasttext,
    create_fasttext_features,
    save_fasttext_features,
    save_fasttext_model
)

df_lemma = pd.read_csv(
    PROCESSED_DATA_DIR / "clean_lemma.csv"
)

df_lemma["text_lemma"] = (
    df_lemma["text_lemma"]
    .fillna("")
    .astype(str)
)

print(df_lemma.shape)
print(df_lemma.columns.tolist())

X_train_lemma, X_test_lemma, y_train_lemma, y_test_lemma = split_data(
    df_lemma,
    "text_lemma"
)

lemma_dir = PROCESSED_DATA_DIR / "neural"

ft_lemma = train_fasttext(
    X_train_lemma,
    corpus_file=str(
        lemma_dir / "train_corpus.txt"
    )
)

save_fasttext_model(
    ft_lemma,
    lemma_dir
)

print("finished")
```

```python
X_train_ft_lemma, X_test_ft_lemma = create_fasttext_features(
    X_train_lemma,
    X_test_lemma,
    ft_lemma
)
save_fasttext_features(
    X_train_ft_lemma,
    X_test_ft_lemma,
    y_train_lemma,
    y_test_lemma,
    output_dir=lemma_dir
)
```
